```python
import pandas as pd
import numpy as np
```

Pandas --- Data Analysis(EDA), Data Manipulations, Data Cleaning(Pre Processing)

A DataFrame is a labeled 2-dimensional data structure used to store and manipulate tabular data — basically a programmable table.
DataFrame is a specialized version of Dictionary DataStructure in python.

```python
# Creating Empty Dataframes
df = pd.DataFrame()
print(df)
```

```python
# Creating Dataframes from the list

lst = ['Geeks', 'For', 'Geeks', 'is', 'portal', 'for', 'Geeks']
df = pd.DataFrame(lst)
print(df)
```

```python
# Creating Dataframes from numpy array list

data = { 'A': np.array([1, 4, 7]),
         'B': np.array([2, 5, 8]),
         'C': np.array([3, 6, 9]) }
df = pd.DataFrame(data)
print(df)
```

```python
# Creating dataframes from python dictionaries

data = [{
    'a': 1,
    'b': 2
}, {
    'a': 5,
    'b': 10,
    'c': 20
}]
df = pd.DataFrame(data)
print(df)

data = {'Name': ['Jake', 'Eve', 'Charlie'],
        'Age': [ 22, 35, 28],
        'Gender': [ 'Male', 'Female', 'Male'],
        'Salary': [40000, 70000, 48000]}
df = pd.DataFrame(data)
print(df)

# Setting new index
res = df.set_index('Name')

# resetting index
res = df.reset_index(drop=True)
print(res)
```

### Indexing with loc
##### allows to access the rows and columns of the dataframe using their labels

```python
data = {'age': [25, 30], 'city': ['NY', 'LA']}
df = pd.DataFrame(data, index=['Alice', 'Bob'])
row = df.loc['Alice']
print(df)
print("")
print(row)
```

### set_index()
##### setting the column as the index instead of using the default index 0,1,2,3......

```python
data = {
    "ID": [101, 102, 103],
    "Name": ["Rahul", "Amit", "Vikram"],
    "Age": [25, 30, 28]
}

df = pd.DataFrame(data)
print(df)

df = df.set_index("ID")
print(df)
```

### setting up multi index

```python
df = pd.DataFrame({
    "Department": ["IT", "IT", "HR"],
    "Employee_ID": [101, 102, 103],
    "Name": ["Rahul", "Amit", "Vikram"]
})

df.set_index(["Department", "Employee_ID"], inplace=True)
print(df)
```

```python
## resetting the index to default

df.reset_index(inplace=True)
print(df)
```

### indexing with loc
##### loc[] is used to access rows and columns of a DataFrame using their labels (index names and column names).

```python
# Accessing a single row
print(df.loc[0])
print("")

# Accessing multiple rows
print(df.loc[[0, 2]])
print("")

# Accessing a single specific element
print(df.loc[0, 'Department'])
print("")

# Slicing Rows with the loc
print(df.loc[0:2])
print("")

# Filtering rows with the condition
print(df.loc[df['Department'] == "IT"])
```

# Concatenate


```python
df1 = pd.DataFrame({
    'a': [122.0, 118.0],
    'b': [0, 0],
    'c': [23, 54]
})


df2 = pd.DataFrame({
    'a': [-122.0, -118.0],
    'b': [234, 564]
})

concatenated_df = pd.concat([df1, df2], ignore_index=True, axis=1)
print(concatenated_df.tail())
```

# Merging
##### pd.merge()
##### similar as inner join in sql

```python
users = pd.DataFrame({'userid':[1,2,3], 'name':['A','B','C']})
msgs = pd.DataFrame({'userid':[1,1,2], 'msg':['hello', 'bye', 'hi']})

# must have one common column like here is userid
pd.merge(users, msgs, on='userid')
```
